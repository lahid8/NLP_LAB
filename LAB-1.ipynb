Pre-Lab Task
1. Define Text Cleaning.
Text cleaning is the process of removing unwanted characters, HTML tags, URLs, digits, and extra spaces from text to make it suitable for NLP tasks.






3. Methods used to remove unwanted elements
Element	Method
HTML tags	BeautifulSoup or re.sub()
URLs	re.sub(r'https?://\S+', '', text)
Email addresses	re.sub(r'\S+@\S+', '', text)
Digits	re.sub(r'\d+', '', text)
Extra spaces	re.sub(r'\s+', ' ', text).strip()
Uppercase letters	text.lower()




4. What is the purpose of regular expressions?
Regular expressions are used to search, match, and replace patterns in text.








6. Which Python library is used to remove HTML tags?
The BeautifulSoup class from the bs4 library.








7. What is the difference between \d+ and \W+?
- \d+ matches one or more digits, such as 123.
- \W+ matches one or more non-word characters, such as punctuation and symbols. It also matches spaces.

In- Lab Task
Tasks
1. Create a Python list containing the given text corpus.
2. Import the required libraries (re and BeautifulSoup).
3. Convert all text to lowercase.
4. Remove HTML tags.
5. Remove URLs.
6. Remove email addresses.
7. Remove digits.
8. Remove punctuation and special characters.
9. Remove extra white spaces.
10. Store the cleaned text in a new list named cleaned_corpus.
11. Display the original corpus and the cleaned corpus.
12. Compare the text before and after cleaning.

```python

import re
from bs4 import BeautifulSoup

corpus = [
    "<p>Data Science is AMAZING!!! Visit https://datascience.com</p>",
    "Python 3.12 is released in 2025.",
    "NLP is used in AI.",
    "Contact us at info@gmail.com"
]

cleaned_corpus = []

for text in corpus:
    text = BeautifulSoup(text, "html.parser").get_text()
    text = text.lower()
    text = re.sub(r'https?://\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    cleaned_corpus.append(text)

print("Original Corpus:")
for text in corpus:
    print(text)

print("\nCleaned Corpus:")
for text in cleaned_corpus:
    print(text)
```
